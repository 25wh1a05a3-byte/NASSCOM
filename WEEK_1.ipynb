# Writing and Using Functions in Python


Defining and calling functions

Positional and default arguments

*args (variable arguments)

**kwargs (keyword arguments)



```python
def greet():
    print("Hello! Welcome to Python Functions.")

greet()
```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")
```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
from functools import reduce


input_string = input("Enter numbers separated by spaces: ")
numbers = list(map(int, input_string.split()))

squared_numbers = list(map(lambda x: x**2, numbers))

even_squares = list(filter(lambda x: x % 2 == 0, squared_numbers))

total_sum = reduce(lambda x, y: x + y, even_squares, 0)

print("Original Numbers:", numbers)
print("Squared Numbers :", squared_numbers)
print("Even Squares    :", even_squares)
print("Sum of Squares  :", total_sum)
```

```python
prices = {"laptop": 1000, "mouse": 50, "monitor": 300}

premium_discounted = dict(
    map(lambda x: (x[0], x[1] * 0.8),
        filter(lambda x: x[1] > 100, prices.items()))
)

print(premium_discounted)
```

```python
fruits = ['apple', 'banana', 'cherry', 'apple']
print("Original list:", fruits)
print("First fruit:", fruits[0])


```

```python
fruits.append('orange')
print("After appending:", fruits)

```

```python
fruits.remove('banana')
print("After removing banana:", fruits)

print("Number of fruits:", len(fruits))
```

```python
dimensions = (10, 20, 30)
print("Tuple:", dimensions)

print("Second element:", dimensions[1])
```

```python
unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)

unique_numbers.add(4)
print("After adding 4:", unique_numbers)

unique_numbers.remove(2)
print("After removing 2:", unique_numbers)
```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

```python
student = {
    'name': 'Alice',
    'age': 21,
    'course': 'Python'
}
print("Dictionary:", student)

print("Student name:", student['name'])
```

```python
student['grade'] = 'A'
print("After adding grade:", student)

del student['age']
print("After deleting age:", student)

for key, value in student.items():
    print(f"{key}: {value}")
```
